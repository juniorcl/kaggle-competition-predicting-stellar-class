# Imports

```python
import numpy as np
import pandas as pd

from sklearn.preprocessing import LabelEncoder
```

# Loading Dataset

```python
X_train_raw = pd.read_parquet('../data/X_train_raw.parquet')
y_train = pd.read_parquet('../data/y_train.parquet')

X_test_raw = pd.read_parquet('../data/X_test_raw.parquet')
```

# Feature Engineering

## Category Type

```python
X_train_raw['spectral_type'] = X_train_raw['spectral_type'].astype('category')
X_train_raw['galaxy_population'] = X_train_raw['galaxy_population'].astype('category')

X_test_raw['spectral_type'] = X_test_raw['spectral_type'].astype('category')
X_test_raw['galaxy_population'] = X_test_raw['galaxy_population'].astype('category')
```

### u subtract g and g subtract r and r subtract i and i subtract z

```python
X_train_fe = X_train_raw.copy()
X_test_fe = X_test_raw.copy()

X_train_fe['u_sub_g'] = X_train_fe['u'] - X_train_fe['g']
X_test_fe['u_sub_g'] = X_test_fe['u'] - X_test_fe['g']

X_train_fe['g_sub_r'] = X_train_fe['g'] - X_train_fe['r']
X_test_fe['g_sub_r'] = X_test_fe['g'] - X_test_fe['r']

X_train_fe['r_sub_i'] = X_train_fe['r'] - X_train_fe['i']
X_test_fe['r_sub_i'] = X_test_fe['r'] - X_test_fe['i']

X_train_fe['i_sub_z'] = X_train_fe['i'] - X_train_fe['z']
X_test_fe['i_sub_z'] = X_test_fe['i'] - X_test_fe['z']
```

### u subtract r and u subtract z

```python
X_train_fe['u_sub_r'] = X_train_fe['u'] - X_train_fe['r']
X_test_fe['u_sub_r'] = X_test_fe['u'] - X_test_fe['r']

X_train_fe['u_sub_z'] = X_train_fe['u'] - X_train_fe['z']
X_test_fe['u_sub_z'] = X_test_fe['u'] - X_test_fe['z']
```

### redshift mult r and redshift mult i

```python
X_train_fe['redshift_mult_r'] = X_train_fe['redshift'] * X_train_fe['r']
X_test_fe['redshift_mult_r'] = X_test_fe['redshift'] * X_test_fe['r']

X_train_fe['redshift_mult_i'] = X_train_fe['redshift'] * X_train_fe['i']
X_test_fe['redshift_mult_i'] = X_test_fe['redshift'] * X_test_fe['i']
```

### Spatial Coordinates

```python
X_train_fe['x_espatial_coordinates'] = np.cos(X_train_fe['delta']) * np.cos(X_train_fe['alpha'])
X_test_fe['x_espatial_coordinates'] = np.cos(X_test_fe['delta']) * np.cos(X_test_fe['alpha'])
```

```python
X_train_fe['y_espatial_coordinates'] = np.cos(X_train_fe['delta']) * np.sin(X_train_fe['alpha'])
X_test_fe['y_espatial_coordinates'] = np.cos(X_test_fe['delta']) * np.sin(X_test_fe['alpha'])
```

```python
X_train_fe['z_espatial_coordinates'] = np.sin(X_train_fe['delta'])
X_test_fe['z_espatial_coordinates'] = np.sin(X_test_fe['delta'])
```
